```python
age_corr = df.corr()["age"].drop("age")
print(age_corr.sort_values(ascending=False))
best_feature = age_corr.sort_values(ascending=False). index[0]
print("Best single Feature for predicting Age:", best_feature)
```

```python
from sklearn.model_selection import train_test_split
X = df.drop(columns=["DEATH_EVENT"])
Y = df["DEATH_EVENT"]
X_train, X_test, Y_train, Y_test, = train_test_split(X,Y, test_size=0.2, random_state=42) 
```

```python
print(X_train.shape)
print(X_test.shape)
print(Y_train.shape)
print(Y_train.shape)
```

```python
from sklearn.linear_model import LinearRegression
from sklearn.metrics import mean_absolute_error, mean_squared_error, r2_score
X1 = df[[best_feature]]
Y_age = df["age"]
X1_train, X1_test, Y1_train, Y1_test = train_test_split(X1, Y_age, test_size=0.2, random_state=42)
model1 = LinearRegression()
model1.fit(X1_train, Y1_train)
pred1 = model1.predict(X1_test)
print("Model 1 MAE:", mean_absolute_error(Y1_test, pred1))
print("Model 1 MSE:", mean_squared_error(Y1_test, pred1))
print("Model 1 R2:", r2_score(Y1_test, pred1)) 
```

```python
top2 = age_corr.abs(). sort_values(ascending=False).index[:2]
X2 = df[list(top2)]
Y_age = df["age"]
X2_train, X2_test, Y2_train, Y2_test = train_test_split(X2, Y_age, test_size= 0.2, random_state=42)
model2 = LinearRegression()
model2.fit(X2_train, Y2_train)
pred2 = model2.predict(X2_test)
print("Model 2 MAE:", mean_absolute_error(Y2_test, pred2))
print("Model 2 MSE:", mean_squared_error(Y2_test, pred2))
print("Model 2 R2:", r2_score(Y2_test, pred2)) 


```

```python
X_all = df.drop(columns = ["age"])
Y_age = df["age"]
Xa_train, Xa_test, Ya_train, Ya_test = train_test_split(X_all, Y_age, test_size = 0.2, random_state =42)
model3 = LinearRegression()
model3.fit(Xa_train, Ya_train)
pred3 = model3.predict(Xa_test)
print("Model 3 MAE:", mean_absolute_error(Ya_test, pred3))
print("Model 3 MSE:", mean_squared_error(Ya_test, pred3))
print("Model 3 R2:", r2_score(Ya_test, pred3)) 
```

```python
from sklearn.linear_model import LogisticRegression
from sklearn.metrics import classification_report, confusion_matrix
log_model = LogisticRegression(max_iter=1000)
log_model.fit(X_train, Y_train)
log_pred = log_model.predict(X_test)
print("=== Logistic Regression Report ===")
print(classification_report(Y_test, log_pred))
print(confusion_matrix(Y_test, log_pred))
```
